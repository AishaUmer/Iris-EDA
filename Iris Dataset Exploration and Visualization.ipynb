# Introduction

The Iris dataset is one of the most famous and foundational datasets in the fields of statistics and machine learning. Introduced by the British statistician and biologist Ronald Fisher in his 1936 paper, it serves as a classic benchmark for testing data exploration techniques and classification algorithms.

# Problem Statement

When dealing with structured data, jumping straight into complex machine learning models without understanding the underlying data distribution can lead to poor results. The goal of this project is to perform Exploratory Data Analysis (EDA) on the Iris dataset using pandas, matplotlib, and seaborn to analyze the structural patterns, distributions, and relationships between the physical traits of different iris flower species.

# Dataset Understanding and Description

The Iris dataset consists of 150 rows and 6 columns of records. It records specific morphometric measurements of three closely related iris species.

Sepal Length (Continuous numeric, in cm)

Sepal Width (Continuous numeric, in cm)

Petal Length (Continuous numeric, in cm)

Petal Width (Continuous numeric, in cm)

Species (Categorical target variable: Setosa, Versicolor, Virginica)

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

# Data Loading and Analysis

```python
df = pd.read_csv('Iris.csv')

print(df.to_string())
```

# Visualization

```python
df.plot(kind = "scatter" , x = "SepalLengthCm" , y = "PetalLengthCm")

plt.show()
```

```python
df.plot(kind = "scatter" , x = "SepalLengthCm" , y = "SepalWidthCm")

plt.show()
```

```python
df.plot(kind = "scatter" , x = "PetalLengthCm" , y = "PetalWidthCm")

plt.show()
```

```python
plt.hist(df['PetalLengthCm'], bins=20)
plt.xlabel('Petal Length')
plt.ylabel('Frequency')
plt.title('Petal Length Distribution')
plt.show()
```

```python
sns.histplot(data=df, x='SepalLengthCm')

plt.title('Distribution of Sepal Length by Species')
plt.show()
```

```python
sns.boxplot(x="Species", y="PetalLengthCm", data=df )
plt.show()
```

```python
sns.boxplot(x="Species", y="SepalLengthCm", data=df )
plt.show()
```

```python
sns.boxplot(x="Species", y="SepalWidthCm", data=df )
plt.show()
```

# Conclusion

Based on the visualizations:

Iris Setosa is Highly Distinct: The Iris-setosa species is completely linearly separable from the other two species, particularly when looking at Petal Length and Petal Width. It features significantly smaller petals but generally wider sepals.

Iris Versicolor and Virginica Overlap: Iris-versicolor and Iris-virginica share similar characteristics and show a slight overlap in their distributions. However, Virginica consistently trends larger in both petal length and width than Versicolor.

Petal Measurements are Better Predictors: The scatter and box plots reveal that petal measurements have a much tighter correlation with the specific species type than sepal measurements.

Outlier Detection: The box plots indicate that the dataset is clean, with very few outliers mostly confined to a few unusual sepal width measurements in the Versicolor or Setosa groups.
